```python
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 1 Import the Data
df = pd.read_csv('medical_examination.csv')
df
```

```python
# Add an overweight column to the data
# IMC = peso(kg) / altura(m)**2 . Si > 25 es 1(True) , sino 0(False)
df['overweight'] = (df['weight'] / (df['height'] /100)**2 >25 ).astype('int8')
df['overweight'].head()
```

```python
# Normalize data
df['cholesterol'] = (df['cholesterol'] > 1).astype('int8')
df['gluc'] = (df['gluc'] > 1).astype('int8')
```

```python
# Draw the Categorical Plot using pd.melt
df_cat = pd.melt(df , id_vars=['cardio'], value_vars=['cholesterol', 'gluc', 'smoke', 'alco', 'active','overweight'])
df_cat
```

```python
# Group and reformat the data in df_cat
df_cat = df_cat.groupby(['cardio', 'variable', 'value']).size().reset_index(name='total')
df_cat
```

```python
# Making the graphic
g = sns.catplot(x='variable', y='total', hue='value',col='cardio', data=df_cat,kind='bar')
# Obtener the pic
fig = g.figure
```

```python
# Draw the Heat Map 
# Clean the data in the 
df_heat = df[(df['ap_lo'] <= df['ap_hi']) & (df['height'] >= df['height'].quantile(0.025)) & (df['height'] <= df['height'].quantile(0.975))
             & (df['weight'] >= df['weight'].quantile(0.025)) & (df['weight'] <= df['weight'].quantile(0.975))]
df_heat
```

```python
# Calculate the correlation matrix 
corr = df_heat.corr()
corr
```

```python
# Generate a mask for the upper triangle
mask = np.triu(np.ones_like(corr, dtype=bool))
```

```python
# Set up the matplotlib figure.
fig, ax =plt.subplots(figsize=(12,12))
sns.heatmap(corr, mask=mask, annot=True, fmt='.1f', center=0, vmin=-0.1, vmax=0.3, square=True, linewidths=.5, cbar_kws={'shrink':.5},ax=ax)
```

**What methods are best for identifying outliers?**

Outliers can be identified using methods such as boxplots, the interquartile range (IQR), and descriptive statistics. These methods can help identify unusually high or low values in variables such as fare amount, trip distance, tip amount, and total amount.

**How do you make the decision to keep or exclude outliers from any future models?**

#### Scatter plot

**Plot mean trip distance by drop-off location**

```python
# Get number of unique drop-off location IDs
df['DOLocationID'].value_counts()
```

```python
# Calculate the mean trip distance for each drop-off location
mean_distance_by_dropoff = df.groupby('DOLocationID')['trip_distance'].mean()

# Sort the results in descending order by mean trip distance
mean_distance_by_dropoff = mean_distance_by_dropoff.sort_values(ascending=False)
mean_distance_by_dropoff
```

## BONUS CONTENT

To confirm your conclusion, consider the following experiment:
1. Create a sample of coordinates from a normal distribution&mdash;in this case 1,500 pairs of points from a normal distribution with a mean of 10 and a standard deviation of 5
2. Calculate the distance between each pair of coordinates 
3. Group the coordinates by endpoint and calculate the mean distance between that endpoint and all other points it was paired with
4. Plot the mean distance for each unique endpoint

```python
#BONUS CONTENT


# 1. Generate random points on a 2D plane from a normal distribution
points = np.random.normal(loc=10, scale=5, size=(3000, 2))

# 2. Calculate Euclidean distances between points in first half and second half of array
distances = np.sqrt(
    (points[:1500, 0] - points[1500:, 0])**2 +
    (points[:1500, 1] - points[1500:, 1])**2
)

# 3. Group the coordinates by "drop-off location", compute mean distance
dropoff_locations = points[1500:, 0]
mean_distance = pd.DataFrame({
    'DOLocationID': dropoff_locations,
    'distance': distances
}).groupby('DOLocationID')['distance'].mean()

# 4. Plot the mean distance between each endpoint ("drop-off location") and all points it connected to
plt.figure(figsize=(14, 6))

sns.scatterplot(
    x=mean_distance.index,
    y=mean_distance.values
)

plt.title('Mean Distance by Drop-off Location')
plt.xlabel('Drop-off Location')
plt.ylabel('Mean Distance')

plt.show()
```

**Histogram of rides by drop-off location**

First, check to whether the drop-off locations IDs are consecutively numbered. For instance, does it go 1, 2, 3, 4..., or are some numbers missing (e.g., 1, 3, 4...). If numbers aren't all consecutive, the histogram will look like some locations have very few or no rides when in reality there's no bar because there's no location. 

```python
# Check if all drop-off locations are consecutively numbered
df['DOLocationID'].sort_values().diff().dropna().eq(1).all()
```

To eliminate the spaces in the historgram that these missing numbers would create, sort the unique drop-off location values, then convert them to strings. This will make the histplot function display all bars directly next to each other. 

```python
# Sort the unique drop-off location values
dropoff_locations = df['DOLocationID'].unique()

# DOLocationID column is numeric, so sort in ascending order
dropoff_locations = np.sort(dropoff_locations)

# Convert to string
dropoff_locations = dropoff_locations.astype(str)

# Plot
sns.histplot(
    data=df,
    x='DOLocationID',
    discrete=True,
    bins=len(dropoff_locations)
)

plt.title('Distribution of Drop-off Locations')
plt.xlabel('Drop-off Location ID')
plt.ylabel('Count')
plt.show()
```
